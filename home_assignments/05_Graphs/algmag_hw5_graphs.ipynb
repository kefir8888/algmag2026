Мы заводим начальную перестановку 123 и набор операций (генераторов): перестановка стощего на первом месте со стоящим на втором и то же самое со второй позицией и третьей.

Применяем ко всем уже обнаруженным состояниям все генераторы по очереди. Если было найдено хотя бы одно новое состояние, потенциально может быть найдено какое-то, которое достижимо только через него, так что в этом случае процесс продолжается. Если же ни одного нового состояния найдено не было, то все достижимые состояния уже добавлены в список.

Попробуйте добавить 4 в лист states. Прежде чем запустить код, подумайте, сколько будет достижимых состояний.

Попробуйте добавить третий генератор. Сколько теперь будет состояний?

Добавите генератор (1, 3). Изменилось ли число состояний? Почему?

```python
def swap(state, op):
    new_state = state.copy()
    
    new_state[op[0]], new_state[op[1]] = state[op[1]], state[op[0]]

    return new_state

generators = [(0, 1), (1, 2)]

states = [[1, 2, 3]]

found = True

while(found):
    found = False
    
    for state in states:
        for gen in generators:
            new_state = swap(state, gen)

            #print("new, states", new_state, states)

            if (not(new_state in states)):
                #print("found")
                found = True
                states += [new_state]

print(f"{len(states)} states:")
print(states)
```
