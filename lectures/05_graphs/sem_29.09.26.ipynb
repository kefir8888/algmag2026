```python
# aabbababb -> aba

def simplify(s, identities):
    found = True

    #s = identities[-2] + s + identities[-1]
    
    while (found):
        found = False
        
        for id in identities:
            pos = s.find(id)

            if (pos != -1):
                found = True
            
                s = s[:pos] + s[pos + len(id):]

                print(s)

    #if (s == "bab"):
    #    s = "aba"
    
    return s

#s = "aabbababbaababbabaabaababa"
s = "baba"

identities = ["aa", "bb", "ababab", "bababa"]

res = simplify(s, identities)

print("result = ", res)
```

```python
def swap(state, op):
    new_state = state.copy()
    
    new_state[op[0]], new_state[op[1]] = state[op[1]], state[op[0]]

    return new_state

generators = [(0, 1), (1, 2)]

states = [[1, 2, 3]]

found = True

while(found):
    found = False
    
    for state in states:
        for gen in generators:
            new_state = swap(state, gen)

            #print("new, states", new_state, states)

            if (not(new_state in states)):
                #print("found")
                found = True
                states += [new_state]

print(states)
```
